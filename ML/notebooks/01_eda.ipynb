# HealthLink EDA

Exploratory Data Analysis on:
- `patients.csv` (patient_id, name, sex, dob, insurance)
- `slots.csv` (slot_id, appointment_date, appointment_time, is_available)
- `appointments.csv` (appointment_id, slot_id, scheduling_date, appointment_date, appointment_time, scheduling_interval, status, check_in_time, appointment_duration, start_time, end_time, waiting_time, patient_id, sex, age, age_group)

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

appts = pd.read_csv('../data/raw/appointments.csv')
patients = pd.read_csv('../data/raw/patients.csv')
slots = pd.read_csv('../data/raw/slots.csv')

print('Appointments:', appts.shape)
print('Patients:', patients.shape)
print('Slots:', slots.shape)
```

```
# No-show rate
no_show_rate = (appts['status'].str.lower() == 'did not attend').mean()
print(f'No-show rate: {no_show_rate:.2%}')
```

```
# Wait time distribution (attended only)
attended = appts[appts['status'].str.lower() == 'attended'].dropna(subset=['waiting_time'])
wait_min = attended['waiting_time'] / 60
wait_min.hist(bins=50, figsize=(10,4))
plt.xlabel('Wait time (minutes)')
plt.title('Appointment Wait Time Distribution')
plt.show()
```

```
# No-show by age group
no_show_by_age = appts.groupby('age_group')['status'].apply(
    lambda x: (x.str.lower() == 'did not attend').mean()
).sort_values(ascending=False)
no_show_by_age.plot(kind='bar', figsize=(12,4))
plt.title('No-show rate by age group')
plt.ylabel('No-show rate')
plt.show()
```

```
# Scheduling interval vs no-show
appts['no_show'] = (appts['status'].str.lower() == 'did not attend').astype(int)
interval_noshow = appts.groupby('scheduling_interval')['no_show'].mean()
interval_noshow[:30].plot(figsize=(12,4))
plt.title('No-show rate by scheduling interval (days ahead)')
plt.ylabel('No-show rate')
plt.show()
```
